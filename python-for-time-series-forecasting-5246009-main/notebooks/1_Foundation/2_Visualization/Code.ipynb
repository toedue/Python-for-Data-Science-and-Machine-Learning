# Visualize time series data

## Load data

```python
path = '../../../data/FRED/FRED_joined.parquet'
```

```python
import pandas as pd
df = pd.read_parquet(path)

df
```

For demo purposes, we'll use one of the columns in the dataset for this section.

```python
s = df['CORESTICKM159SFRBATL']
s
```

## Plotting libraries

```python
plt.plot(x, y)
```

## Customizing the plot

### Change the title and size

```python
s.plot(title='CORESTICKM159SFRBATL', width=1000, height=500)
```

### Legend off

```python
fig = s.plot(title='CORESTICKM159SFRBATL', width=1000, height=500)
fig.update_layout(showlegend=False)
```

### Theme template

```python
s.plot(title='CORESTICKM159SFRBATL', width=1000, height=500, template='plotly_dark')
```

### Default theme to plotly_dark

```python
import plotly.io as pio
pio.templates.default = "plotly_dark"
```

```python
s.plot()
```

### Update template

```python
pio.templates["plotly_dark"].layout.update(
    width=1000,
    height=500,
    showlegend=False,
    autosize=False
)
```

```python
s.plot()
```
