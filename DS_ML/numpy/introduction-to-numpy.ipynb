```python
# Standard Deviation
# Standard deviation (SD) measures how far the data values are from the mean (average).
np.std(a2)
```

```python
# Variance is a measure of how spread out the values in a dataset are from their mean (average).
# Small variance → the values are close to the mean.
# Large variance → the values are far from the mean.
np.var(a2)
```

```python
np.sqrt(np.var(a2))
```
