# Support Vector Machine (Classification & Regression)

Linear Kernel K(x, xi) = sum(x * xi)

Polynomial Kernel K(x,xi) = 1 + sum(x * xi)^d

Radial Basis K(x,xi) = exp(-gamma*sum((x-xi^2))

```python
x, y = datasets.load_breast_cancer(return_X_y=True)

x.shape, y.shape
```

```python
x_train, x_test, y_trian, y_test = model_selection.train_test_split(x, y, test_size=0.1, random_state=42, stratify=y)
x_train.shape, x_test.shape, y_trian.shape, y_test.shape
```

```python
clf = svm.SVC(kernel="linear")
clf.fit(x_train, y_trian)
```

```python
y_predict = clf.predict(x_test)
y_predict
```

```python
metrics.accuracy_score(y_test,y_predict)
```

```python
x, y = datasets.fetch_california_housing(return_X_y=True)
x.shape, y.shape
```

```python
y_mean = y.mean()
y_std = y.std()

y_norm = (y-y_mean)/y_std

x_mean = x.mean()
x_std = x.std()

x_norm = (x-x_mean)/x_std
```

```python
x_train, x_test, y_trian, y_test = model_selection.train_test_split(x_norm, y_norm, test_size=0.2, random_state=42)
x_train.shape, x_test.shape, y_trian.shape, y_test.shape
```

```python
model = svm.SVR()
model.fit(x_train, y_trian)
y_predict = model.predict(x_test)
```

```python
metrics.mean_squared_error(y_test,y_predict)
```

```python
metrics.r2_score(y_test,y_predict)
```
