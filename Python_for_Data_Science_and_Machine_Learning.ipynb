## Functional API

```python
import tensorflow as tf
import matplotlib.pyplot as plt
import pandas as pd
```

```python
(x_train, y_train) , (x_test, y_test) = tf.keras.datasets.mnist.load_data()

x_train.shape, y_train.shape,x_test.shape, y_test.shape
```

```python
plt.imshow(x_train[4], cmap="gray")
```

```python
x_train, x_test = x_train.reshape(-1, 28*28), x_test.reshape(-1, 28*28)

x_train, x_test = x_train/255, x_test/255
```

```python
y_train, y_test = tf.one_hot(y_train, depth=10), tf.one_hot(y_test, depth=10)
```

```python
def wide_block(x, hidden_unit=512, dropout_rate=0.2):
  hidden1 = tf.keras.layers.Dense(units=hidden_unit, activation=tf.nn.leaky_relu)(x)
  hidden1 = tf.keras.layers.Dropout(dropout_rate)(hidden1)
  hidden1 = tf.keras.layers.Dense(units=hidden_unit, activation=tf.nn.leaky_relu)(hidden1)
  hidden1 = tf.keras.layers.Dropout(dropout_rate)(hidden1)
  hidden2 = tf.keras.layers.Dense(units=hidden_unit, activation=tf.nn.leaky_relu)(hidden1)
  concat = tf.keras.layers.Concatenate()([input_layer, hidden2])
  return concat
```

```python
tf.keras.utils.plot_model(model)
```

```python
model.compile('adam',
              loss='categorical_crossentropy',
              metrics=['acc'])
```

```python
model.fit(x=x_train, y=y_train, batch_size=32, epochs=10, validation_split=0.1)
```

```python
df = pd.DataFrame(model.history.history)
df.plot(figsize=(10,8))
```

## Subclassing API

```python
class WideDeepModel(tf.keras.Model):
    def __init__(self, input_shape, units=30, act='relu', **kwargs):
        super().__init__(**kwargs)
        self.input_dim = input_shape  # store it, don't build an Input here
        self.hidden1 = tf.keras.layers.Dense(units, activation=act)
        self.dropout1 = tf.keras.layers.Dropout(0.2)
        self.hidden2 = tf.keras.layers.Dense(units, activation=act)
        self.dropout2 = tf.keras.layers.Dropout(0.2)
        self.out = tf.keras.layers.Dense(10, activation='softmax')

    def call(self, inputs, training=False):
        x = self.hidden1(inputs)
        x = self.dropout1(x, training=training)
        x = self.hidden2(x)
        x = self.dropout2(x, training=training)
        concat = tf.keras.layers.concatenate([x, inputs])  # wide + deep
        return self.out(concat)

    def build_graph(self):
        x = tf.keras.layers.Input(self.input_dim)
        return tf.keras.Model(inputs=x, outputs=self.call(x))

```

```python
model = WideDeepModel(input_shape=(784,))
model.build(input_shape=(None, 784))
```

```python
model.summary()
```

```python
tf.keras.utils.plot_model(model.build_graph())
```

```python
model.compile('adam',
              loss='categorical_crossentropy',
              metrics=['acc'])

model.fit(x=x_train, y=y_train, batch_size=32, epochs=10, validation_split=0.1)
```

```python
df = pd.DataFrame(model.history.history)
df.plot(figsize=(10,8))
```
